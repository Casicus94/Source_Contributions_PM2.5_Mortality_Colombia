```python
for ifile,file in enumerate(files_pm25):
    print(f'--------------> PM2_5_{cat[ifile]}_ymean.nc')
    if os.path.exists(f'{path}{cat[ifile]}/PM2_5_{cat[ifile]}_ymean.nc'):
        continue
    ds = xr.open_dataset(file)
    ##Monthly mean
    monmean = ds.PM2_5_DRY.resample(XTIME='ME').mean()
    monmean.to_netcdf(f'{path}{cat[ifile]}/PM2_5_{cat[ifile]}_ymonmean.nc')
    ##Yearly max
    ymean = monmean.mean(dim=['XTIME'])
    ymean.to_netcdf(f'{path}{cat[ifile]}/PM2_5_{cat[ifile]}_ymean.nc')
```
